````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型
# ============================================================
!pip install -q groq gradio

import sqlite3
import pandas as pd
from IPython.display import display
from google.colab import userdata
from groq import Groq
import gradio as gr

# 🔴 金鑰只能從左側「🔑 密鑰（Secrets）」讀取，絕對不可以寫成 api_key="gsk_..."
#    寫死的金鑰一旦存到 GitHub，就等於公開給全世界使用。
#    也不要 print 出金鑰，因為輸出結果也會一起存到 GitHub。
try:
    client = Groq(api_key=userdata.get('SDDA_Lectrue_05'))
except Exception as e:
    print("❌ 讀不到金鑰。請檢查：")
    print("   1. 左側 🔑 是否新增了名稱為 SDDA_Lectrue_05 的密鑰（大小寫要完全一樣）")
    print("   2. 該密鑰的「筆記本存取權」開關是否已打開")
    raise

# 模型名稱集中放在這裡；日後模型更換或下架，只要改這一行
MODEL = "openai/gpt-oss-120b"
print("✅ SDDA_Lectrue_05 金鑰讀取成功，使用模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 說明：每次執行先刪除舊表 (DROP TABLE)，重建資料表並寫入初始資料
# ============================================================

def init_db():
    # 建立資料庫連線
    conn = sqlite3.connect('clinic.db')
    cursor = conn.cursor()
    try:
        # 每次執行都先刪除舊資料表，確保環境乾淨
        cursor.execute("DROP TABLE IF EXISTS departments")

        # 建立科別與名額表
        cursor.execute("""
            CREATE TABLE departments (
                name TEXT PRIMARY KEY,
                quota INTEGER NOT NULL,
                booked INTEGER NOT NULL
            )
        """)

        # 放入初始資料（內科、外科、牙科）
        initial_data = [
            ('內科', 50, 0),
            ('外科', 40, 0),
            ('牙科', 30, 0)
        ]
        cursor.executemany("INSERT INTO departments (name, quota, booked) VALUES (?, ?, ?)", initial_data)

        conn.commit()
        print("✅ 成功建立資料庫與初始資料！")
    finally:
        conn.close()

# 執行初始化函式
init_db()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 說明：定義 show_tables() 函式並呼叫一次，以 DataFrame 呈現目前名額狀況
# ============================================================

def show_tables():
    conn = sqlite3.connect('clinic.db')
    try:
        # 使用 pandas 讀取資料表內容，並計算剩餘名額
        df = pd.read_sql(
            "SELECT name AS '科別', quota AS '總名額', booked AS '已掛號', (quota - booked) AS '剩餘名額' FROM departments",
            conn
        )
        print("📊 目前診所各科名額狀態（學號：411312545 聶運安）：")
        display(df)
    finally:
        conn.close()

# 呼叫一次以顯示初始狀態
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 說明：呼叫 Groq 模型解析使用者話語，限制 temperature=0 並進行白名單驗證
# ============================================================

def interpret_user_input(text):
    prompt = f"""
你是一個診所掛號系統的語意解析助理。
請從使用者的話語中提取出「動作」與「對象（科別）」。

允許的動作（Action）只能是：掛號、取消、查詢、未知
允許的對象（Department）只能是：內科、外科、牙科、未知

安全規定（極度重要）：
1. 絕對不可以提供任何醫療建議。
2. 如果使用者描述症狀（例如頭痛、發燒、肚子痛、感冒等），但沒有明確說出科別，動作與對象一律回傳「未知,未知」。

請嚴格僅回傳以下格式，不要有其他贅字：
動作,對象

使用者說的話：{text}
"""
    try:
        response = client.chat.completions.create(
            model=MODEL,
            messages=[{"role": "user", "content": prompt}],
            temperature=0
        )
        raw_output = response.choices[0].message.content.strip()
    except Exception as e:
        raw_output = "未知,未知"

    # 解析 AI 回覆
    parts = raw_output.split(',')
    if len(parts) == 2:
        action = parts[0].strip()
        department = parts[1].strip()
    else:
        action, department = "未知", "未知"

    # 白名單驗證：不可直接相信 AI 的回覆
    valid_actions = ["掛號", "取消", "查詢"]
    valid_depts = ["內科", "外科", "牙科"]

    if action not in valid_actions:
        action = "未知"
    if department not in valid_depts:
        department = "未知"

    return action, department, raw_output

print("✅ AI 解讀函式載入完成！")
````

````python
# ============================================================
# 第 5 格：測試清單
# 說明：執行至少 8 句測試清單（含正常、錯誤、邊界），印出 AI 原始回覆與系統解讀
# ============================================================

# 格式：(使用者說的話, 預期動作, 預期科別)
test_cases = [
    ("我要掛內科", "掛號", "內科"),
    ("查詢牙科剩餘名額", "查詢", "牙科"),
    ("取消內科掛號", "取消", "內科"),
    ("外科還有名額嗎", "查詢", "外科"),
    ("我今天發燒感冒", "未知", "未知"),
    ("幫我掛號", "掛號", "未知"),
    ("連續掛滿內科 50 人後再掛一次", "掛號", "內科"),  # 規格對應正常語意解析
    ("在沒有掛號的情況下取消牙科", "取消", "牙科")     # 規格對應正常語意解析
]

def run_tests():
    passed = 0
    print("🧪 開始執行自動化測試清單...\n")
    for i, (text, exp_action, exp_dept) in enumerate(test_cases, 1):
        action, dept, raw = interpret_user_input(text)

        # 比對結果
        match = (action == exp_action) and (dept == exp_dept)
        if match:
            passed += 1
            status = "✅ 通過"
        else:
            status = "❌ 失敗"

        print(f"測試 #{i} {status}")
        print(f"  💬 使用者說：{text}")
        print(f"  🤖 AI 原始回覆：{raw}")
        print(f"  ⚙️ 系統解讀：動作={action}, 科別={dept} (預期：{exp_action}, {exp_dept})")
        print("-" * 50)

    print(f"🎯 測試結果：通過 {passed} / {len(test_cases)} 句")

run_tests()
````

````python
# ============================================================
# 第 6 格：功能函式（掛號、取消、查詢）
# 說明：處理掛號、取消、查詢邏輯，包含 fetchone() 的 None 檢查與 try/finally 關閉連線
# ============================================================

def process_clinic_request(action, department):
    if action == "未知" or department == "未知":
        return "⚠️ 系統無法識別您的需求。如果您有身體不適，請直接告知要掛哪一科（內科、外科、牙科），或請直接洽詢櫃台，本系統不提供醫療建議。"

    conn = sqlite3.connect('clinic.db')
    cursor = conn.cursor()
    try:
        # 使用 ? 參數查詢資料
        cursor.execute("SELECT quota, booked FROM departments WHERE name = ?", (department,))
        row = cursor.fetchone()

        # 檢查 fetchone() 是否為 None（防呆保護）
        if row is None:
            return f"❌ 找不到「{department}」這個科別。"

        quota, booked = row

        if action == "掛號":
            if booked >= quota:
                return f"⚠️ 抱歉，{department} 今日名額已滿（總名額 {quota} 人）。"
            cursor.execute("UPDATE departments SET booked = booked + 1 WHERE name = ?", (department,))
            conn.commit()
            return f"✅ 成功掛號【{department}】！目前已掛號人數：{booked + 1}/{quota}"

        elif action == "取消":
            if booked <= 0:
                return f"⚠️ 查無您的掛號紀錄，無法取消（{department} 目前已掛號人數為 0）。"
            cursor.execute("UPDATE departments SET booked = booked - 1 WHERE name = ?", (department,))
            conn.commit()
            return f"✅ 已成功取消【{department}】掛號。目前已掛號人數：{booked - 1}/{quota}"

        elif action == "查詢":
            remaining = quota - booked
            return f"ℹ️ 【{department}】總名額：{quota} 人，已掛號：{booked} 人，剩餘名額：{remaining} 人。"

        else:
            return "❌ 未知的動作指令。"

    finally:
        conn.close()

print("✅ 功能函式載入完成！")
````

````python
# ============================================================
# 第 7 格：主流程與完整流程測試
# 說明：模擬多個使用者情境，並在最後呼叫 show_tables() 確認資料庫變化
# ============================================================

def simulate_user_utterance(text):
    print(f"🗣️ 使用者說：\"{text}\"")
    action, dept, raw = interpret_user_input(text)
    print(f"   ↳ AI 解析結果：動作【{action}】、科別【{dept}】（原始回覆：{raw}）")
    result_msg = process_clinic_request(action, dept)
    print(f"   ↳ 系統回應：{result_msg}\n")

# 每次執行先重置資料庫
init_db()

print("🚀 開始完整流程測試：\n")
simulate_user_utterance("我要掛內科")
simulate_user_utterance("查詢內科還有多少名額")
simulate_user_utterance("我今天肚子痛很難過")  # 測試安全規定（拒絕醫療建議）
simulate_user_utterance("取消內科掛號")
simulate_user_utterance("在沒有掛號的情況下取消牙科")  # 邊界測試
simulate_user_utterance("我要掛牙科")

# 最後呼叫 show_tables() 確認資料庫有變化
show_tables()
````

````python
# ============================================================
# 第 8 格：Gradio 網頁
# 說明：建立一個簡單的互動式網頁介面供民眾使用掛號系統
# ============================================================

def clinic_chatbot(user_message):
    action, dept, _ = interpret_user_input(user_message)
    response = process_clinic_request(action, dept)
    return response

# 建立 Gradio 介面
demo = gr.Interface(
    fn=clinic_chatbot,
    inputs=gr.Textbox(lines=2, placeholder="請輸入您的需求，例如：我要掛內科、查詢牙科名額..."),
    outputs="text",
    title="🏥 聶運安的智能診所掛號系統 (411312545)",
    description="提供內科、外科、牙科之掛號、取消與名額查詢服務。（本系統嚴格遵守安全規定，不提供任何醫療建議）"
)

# 啟動網頁
if __name__ == "__main__":
    demo.launch(inline=False)
````
